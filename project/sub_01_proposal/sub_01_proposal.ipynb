**Description:**

We use the **FRED Macroeconomic Bundle**, a self-proposed dataset drawn from the Federal Reserve Economic Data (FRED) repository, maintained by the Federal Reserve Bank of St. Louis (https://fred.stlouisfed.org). FRED aggregates official, non-synthetic U.S. macroeconomic and financial time series published by government agencies. Each series can be downloaded as CSV via the public URL pattern `https://fred.stlouisfed.org/graph/fredgraph.csv?id=<SERIES_ID>`.

We combine eight series from January 2000 onward, keyed by date:

- **FEDFUNDS** &ndash; Effective Federal Funds Rate (monthly, %)
- **DGS10** &ndash; 10-Year Treasury Yield (daily, %)
- **DGS2** &ndash; 2-Year Treasury Yield (daily, %)
- **CPIAUCSL** &ndash; Consumer Price Index (monthly)
- **UNRATE** &ndash; Unemployment Rate (monthly, %)
- **SP500** &ndash; S&P 500 Index (daily, level)
- **MORTGAGE30US** &ndash; 30-Year Fixed Mortgage Rate (weekly, %)
- **UMCSENT** &ndash; University of Michigan Consumer Sentiment (monthly)

After resampling to monthly frequency and merging, the dataset has roughly 310 rows and 8 core columns, plus derived columns we will compute during wrangling (year-over-year inflation, 10y&ndash;2y yield spread, NBER-based recession flags).

We chose it because it is real, government-published economic data (satisfying the no-synthetic-data rule), its multi-frequency nature requires meaningful wrangling (resampling, alignment, missing-value handling), and it covers dramatic recent events &ndash; the 2008 financial crisis, 2020 COVID shock, 2022 inflation spike, and 2022&ndash;2024 Fed hiking cycle &ndash; which will make our visualizations immediately informative to audiences interested in finance, monetary policy, and the broader economy.
